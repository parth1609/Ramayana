```python
import spacy

# Load a pre-trained English NLP model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Downloading en_core_web_sm model...")
    spacy.cli.download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")


doc = nlp(text)

filtered_tokens = []
in_bracket = False

for token in doc:
    # Check for opening bracket to start skipping
    if token.text == '[':
        in_bracket = True
        continue
    # Check for closing bracket to stop skipping
    if token.text == ']':
        in_bracket = False
        continue

    # If not inside brackets and not a stop word, and not punctuation (unless it's part of a quote)
    if not in_bracket and not token.is_stop and not token.is_punct:
        filtered_tokens.append(token.text)
    elif token.text == '"' or token.text == "'": # Keep quote marks
        filtered_tokens.append(token.text)

# Join the filtered tokens to form the cleaned sentence
# Note: This is a simplified approach to reconstruct the sentence,
# and might not perfectly handle spacing for complex cases.
cleaned_text_nlp = " ".join(filtered_tokens).replace(" .", ".").replace(" ,", ",").replace(" ?", "?").replace(" !", "!")

# Further regex to clean up potential issues from simple join (like " .")
cleaned_text_nlp = re.sub(r'\s([?.!,"](?:\s|$))', r'\1', cleaned_text_nlp)
# Remove potential leading/trailing quotes or other remaining artifacts
cleaned_text_nlp = cleaned_text_nlp.strip().strip("'").strip('"')


print(cleaned_text_nlp)
```

```python

def clean_text(text):
    if not isinstance(text, str):  # Handle non-string values
        return ""
    doc = nlp(text)
    filtered_tokens = []
    in_bracket = False

    for token in doc:
        if token.text == '[':
            in_bracket = True
            continue
        if token.text == ']':
            in_bracket = False
            continue
        if not in_bracket and not token.is_stop and not token.is_punct:
            filtered_tokens.append(token.text)
        elif token.text == '"' or token.text == "'":
            filtered_tokens.append(token.text)

    cleaned_text_nlp = " ".join(filtered_tokens).replace(" .", ".").replace(" ,", ",").replace(" ?", "?").replace(" !", "!")
    cleaned_text_nlp = re.sub(r'\s([?.!,"\"](?:\s|$))', r'\1', cleaned_text_nlp)
    cleaned_text_nlp = cleaned_text_nlp.strip().strip("'").strip('"')
    return cleaned_text_nlp

df['clean_translation'] = df['translation'].apply(clean_text)
df[['translation', 'clean_translation']].head()
```

```python


def convert_devanagari_to_english(text):
    devanagari_to_english = {
        "०": "0",
        "१": "1",
        "२": "2",
        "३": "3",
        "४": "4",
        "५": "5",
        "६": "6",
        "७": "7",
        "८": "8",
        "९": "9",
    }

    converted_text = ""
    if isinstance(text, str):  # Handle cases where the column might have non-string values (e.g., NaN)
        for char in text:
            if char in devanagari_to_english:
                converted_text += devanagari_to_english[char]
            else:
                converted_text += char
    else:
        converted_text = text  # Return the original value if it's not a string

    return converted_text


```

```python
import pandas as pd
import numpy as np

def fill_nan_incrementally(series):
    filled = []
    last_val = None
    for val in series:
        if pd.isna(val):
            if last_val is not None:
                try:
                    new_val = str(int(last_val) + 1)
                except:
                    new_val = last_val
                filled.append(new_val)
                last_val = new_val
            else:
                filled.append(np.nan)
        else:
            filled.append(val)
            last_val = val
    return pd.Series(filled, index=series.index)

# Apply to each (kanda_name, sarga) group
new_df['verse_num_from_shloka'] = new_df.groupby(['kanda_name', 'sarga'])['verse_num_from_shloka'].transform(fill_nan_incrementally)
```

```python
import spacy

# Load a pre-trained English NLP model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Downloading en_core_web_sm model...")
    spacy.cli.download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")


doc = nlp(text)

filtered_tokens = []
in_bracket = False

for token in doc:
    # Check for opening bracket to start skipping
    if token.text == '[':
        in_bracket = True
        continue
    # Check for closing bracket to stop skipping
    if token.text == ']':
        in_bracket = False
        continue

    # If not inside brackets and not a stop word, and not punctuation (unless it's part of a quote)
    if not in_bracket and not token.is_stop and not token.is_punct:
        filtered_tokens.append(token.text)
    elif token.text == '"' or token.text == "'": # Keep quote marks
        filtered_tokens.append(token.text)

# Join the filtered tokens to form the cleaned sentence
# Note: This is a simplified approach to reconstruct the sentence,
# and might not perfectly handle spacing for complex cases.
cleaned_text_nlp = " ".join(filtered_tokens).replace(" .", ".").replace(" ,", ",").replace(" ?", "?").replace(" !", "!")

# Further regex to clean up potential issues from simple join (like " .")
cleaned_text_nlp = re.sub(r'\s([?.!,"](?:\s|$))', r'\1', cleaned_text_nlp)
# Remove potential leading/trailing quotes or other remaining artifacts
cleaned_text_nlp = cleaned_text_nlp.strip().strip("'").strip('"')


print(cleaned_text_nlp)
```

```python

def clean_text(text):
    if not isinstance(text, str):  # Handle non-string values
        return ""
    doc = nlp(text)
    filtered_tokens = []
    in_bracket = False

    for token in doc:
        if token.text == '[':
            in_bracket = True
            continue
        if token.text == ']':
            in_bracket = False
            continue
        if not in_bracket and not token.is_stop and not token.is_punct:
            filtered_tokens.append(token.text)
        elif token.text == '"' or token.text == "'":
            filtered_tokens.append(token.text)

    cleaned_text_nlp = " ".join(filtered_tokens).replace(" .", ".").replace(" ,", ",").replace(" ?", "?").replace(" !", "!")
    cleaned_text_nlp = re.sub(r'\s([?.!,"\"](?:\s|$))', r'\1', cleaned_text_nlp)
    cleaned_text_nlp = cleaned_text_nlp.strip().strip("'").strip('"')
    return cleaned_text_nlp

df['clean_translation'] = df['translation'].apply(clean_text)
df[['translation', 'clean_translation']].head()
```

```python

verse_ref_pattern = r'([०-९]+-[०-९]+-[०-९]+)'

shlokas_series = df['shloka'].astype(str)


extracted_references_df = shlokas_series.str.extractall(verse_ref_pattern)

# Check if any references were found
if not extracted_references_df.empty:
    # Get the Series of extracted references (it's the first captured group, hence column 0)
    extracted_series = extracted_references_df[0]

    # Get unique references and convert to a list for cleaner display
    unique_references = extracted_series.unique().tolist()

    print("Found the following unique verse-like references in the 'shloka' column:")
    if unique_references:
        for ref in unique_references:
            print(ref)
    else:
        # This case should ideally not be hit if extracted_references_df was not empty,
        # but it's a safeguard.
        print("No unique references found after extraction (this is unexpected if matches were found).")
else:
    print("No verse-like references of the pattern 'X-Y-Z' (Devanagari numerals) found in the 'shloka' column.")

```

```python
# Ensure 'shloka' and 'verse_ref' columns are suitable types
# We assume 'df' is your DataFrame, and 'numpy' (as np) and 'pandas' (as pd) are already imported.
df['shloka'] = df['shloka'].astype(str)
# Convert 'verse_ref' to string as well, handle 'nan' strings by converting to actual NaN
df['verse_ref'] = df['verse_ref'].astype(str).replace('nan', np.nan)

# Define the regex pattern to find Devanagari numerals in X-Y-Z or X-Y format.
# [०-९]+ matches one or more Devanagari digits.
# (?:-[०-९]+)? is an optional non-capturing group for a hyphen followed by digits (handles the third part like -Z).
# The outer parentheses () create a capturing group for the entire pattern.
verse_ref_pattern_extract = r'([०-९]+-[०-९]+(?:-[०-९]+)?)'

# 1. Extract the first embedded reference from 'shloka' into a temporary column.
# expand=False ensures that if there's one capture group, the result is a Series.
df['temp_embedded_ref'] = df['shloka'].str.extract(verse_ref_pattern_extract, expand=False)

# 2. Update 'verse_ref' column ONLY IF 'verse_ref' is currently NaN AND an embedded ref was found.
# This prioritizes existing 'verse_ref' values if they are not NaN.
mask_update_verse_ref = df['verse_ref'].isnull() & df['temp_embedded_ref'].notna()
df.loc[mask_update_verse_ref, 'verse_ref'] = df.loc[mask_update_verse_ref, 'temp_embedded_ref']

# 3. Remove all occurrences of these patterns from the 'shloka' column.
# The same regex pattern is used; str.replace will replace all non-overlapping matches.
df['shloka'] = df['shloka'].str.replace(verse_ref_pattern_extract, '', regex=True)

# 4. Clean up potential extra spaces in 'shloka' after removal.
# Replace multiple consecutive spaces with a single space.
df['shloka'] = df['shloka'].str.replace(r'\s{2,}', ' ', regex=True)
# Remove any leading or trailing whitespace from the 'shloka' column.
df['shloka'] = df['shloka'].str.strip()

# 5. Drop the temporary column as it's no longer needed.
df = df.drop(columns=['temp_embedded_ref'])

print("Processed embedded references in 'shloka' and updated 'verse_ref' where necessary.")
# You can optionally display a few rows where 'verse_ref' was updated:
# print("Rows where 'verse_ref' was updated from 'shloka':")
# print(df[mask_update_verse_ref][['kanda_name', 'sarga', 'shloka', 'verse_ref']].head())
print("-" * 50)

# --- Create Arabic Numeral Verse Reference and Identify Issues ---

# Function to convert Devanagari numeral strings (e.g., '१-२-२७') to Arabic numeral strings (e.g., '1-2-27')
def convert_devanagari_to_arabic_ref(dev_ref):
    if pd.isna(dev_ref):
        return np.nan
    dev_ref_str = str(dev_ref)  # Ensure it's a string
    devanagari_digits = "०१२३४५६७८९"
    arabic_digits = "0123456789"
    translation_table = str.maketrans(devanagari_digits, arabic_digits)
    arabic_ref = dev_ref_str.translate(translation_table)
    # If after translation it's empty, only whitespace, or contains no digits (e.g. if original was just '-'), consider it NaN
    if not arabic_ref.strip() or not any(char.isdigit() for char in arabic_ref):
        return np.nan
    return arabic_ref

# Create the 'verse_number_arabic' column
df['verse_number_arabic'] = df['verse_ref'].apply(convert_devanagari_to_arabic_ref)

# Filter for rows where 'verse_number_arabic' is null
null_verse_arabic_df = df[df['verse_number_arabic'].isnull()]

print("\nRows where 'verse_number_arabic' is null (potential issues):")
if not null_verse_arabic_df.empty:
    # Display relevant columns for rows with null 'verse_number_arabic'
    print(null_verse_arabic_df[['kanda_name', 'sarga', 'verse_ref', 'shloka', 'verse_number_arabic']].head())
else:
    print("No rows found with null 'verse_number_arabic'. All verse references seem to be convertible.")
    
print(f"\nNumber of rows with null 'verse_number_arabic': {len(null_verse_arabic_df)}")

# Display a sample of the DataFrame to see the new column and cleaned 'shloka'
# print("\nSample of DataFrame with 'verse_number_arabic' and cleaned 'shloka':")
# print(df[['kanda_name', 'sarga', 'verse_ref', 'shloka', 'verse_number_arabic']].sample(5))
```
